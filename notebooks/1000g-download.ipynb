```python
import urllib.request
import zipfile
import requests
from tqdm import tqdm
```

```python
os.makedirs("../reference_data/1kgp_phase3", exist_ok=True)
url = "https://ndownloader.figshare.com/files/17838962"
output = "../reference_data/1kgp_phase3/1000G_phase3_common_norel.zip"

with requests.get(url, stream=True) as r:
    total = int(r.headers.get("content-length", 0))
    with open(output, "wb") as f, tqdm(total=total, unit="B", unit_scale=True) as bar:
        for chunk in r.iter_content(chunk_size=2**16):
            f.write(chunk)
            bar.update(len(chunk))

with zipfile.ZipFile(output) as z:
    z.extractall("../reference_data/1kgp_phase3/")

print("Download Finished")
```
